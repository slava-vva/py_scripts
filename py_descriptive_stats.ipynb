```python
import statsmodels.api as sm

# Convert OverTime to numeric
df['OverTime_num'] = df['OverTime'].map({'Yes': 1, 'No': 0})

# Independent variables
X = df[['OverTime_num',
        'EnvironmentSatisfaction',
        'YearsAtCompany',
        'MonthlyIncome',
        'HourlyRate',
        'JobInvolvement']]

```

```python
# Dependent variable
y = df['JobSatisfaction']

# Add constant
X = sm.add_constant(X)

# Fit model
model = sm.OLS(y, X).fit()

# Show summary
print(model.summary())
```
