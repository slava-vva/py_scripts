# Student Data
This notebook analyses data for a class


```python
data = [50,57,47,26,33,27,90,35]
print("Data:", data)
print("Mean:", sum(data)/len(data))
print("Median:", sorted(data)[len(data)//2])
print("Mode:", max(set(data), key=data.count))  
```

```python
import numpy as np
print("Standard Deviation:", np.std(data))
print("Doubled Data:", [x * 2 for x in data])
print("Squared Data:", [x ** 2 for x in data])
print("Data Summary:")
print("Min:", min(data))
print(data * 2)
grades = np.array(data)
print("Grades:", grades * 2)
```

```python
#grades.shape
grades[1:]
study_hours = np.array([2, 3, 1, 4, 5, 2, 6, 3])
student_data = np.array([study_hours, grades, data])
avg_study = student_data[0].mean()
avg_grade = student_data[1].mean()
print("Average Study Hours:", avg_study)
print(f"Average Grade: {avg_grade}")
```
