```python
import pandas as pd
df_employee = pd.read_csv('WA_Fn-UseC_-HR-Employee-Attrition.csv')
```

```python
import matplotlib.pyplot as plt

```

```python
plt.bar(df_employee['Department'], df_employee['DailyRate'])
plt.xlabel('Department')
plt.ylabel('Daily Rate')
plt.title('Daily Rate by Department')
plt.show()
```

avg_salary = df_employee.groupby('Department')['DailyRate'].mean()
print(avg_salary)
plt.bar(avg_salary.index, avg_salary.values)
plt.xlabel('Department')
plt.ylabel('Average Daily Rate')
plt.title('Average Daily Rate by Department')
plt.show()

```python
#plt.bar(df_employee['Department'].value_counts().index, df_employee['Department'].value_counts().values)


avg_dailyrate = df_employee[df_employee['Attrition']=='No'].groupby('Department')['DailyRate'].mean()
print(avg_dailyrate)
bars = plt.bar(avg_dailyrate.index, avg_dailyrate.values, color=['blue', 'orange', 'green'])
plt.bar_label(bars)
plt.xlabel('Department')
plt.ylabel('Average Daily Rate')
plt.title('Average Daily Rate by Department')
plt.xticks(rotation=45)
plt.show()
```
