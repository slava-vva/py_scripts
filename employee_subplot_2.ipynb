```python
import pandas as pd
df_employee = pd.read_csv('WA_Fn-UseC_-HR-Employee-Attrition.csv')
```

```python
import matplotlib.pyplot as plt

```

```python
# df_employee.info()
df_employee.columns
# df_employee.dtypes
# df_employee.head()
# df_employee.describe()
# df_employee.shape
```

```python
hourly_rate = df_employee.groupby('JobRole')['HourlyRate'].mean()
hourly_rate.plot(kind='bar', figsize=(10, 6), color='skyblue')
plt.title('Average Hourly Rate by Job Role')
```

```python
fig = plt.figure(figsize=(12, 6))
plt.hist(df_employee['Age'], bins=20, color='lightcoral', edgecolor='black')
plt.title('Distribution of Employee Ages')
plt.xlabel('Age')
plt.ylabel('Frequency')
fig.show()

fig2 = plt.figure(figsize=(12, 6))
plt.scatter(df_employee['MonthlyIncome'], df_employee['YearsAtCompany'], alpha=0.5, color='seagreen')
plt.title('Monthly Income vs. Years at Company')
plt.xlabel('Monthly Income')
plt.ylabel('Years at Company')
fig2.show()
```

```python
min_salary = df_employee['MonthlyIncome'].min()
max_salary = df_employee['MonthlyIncome'].max()
median_salary = df_employee['MonthlyIncome'].median()
mean_salary = df_employee['MonthlyIncome'].mean()
mode_salary = df_employee['MonthlyIncome'].mode()[0]
plt.figure(figsize=(10, 6))
plt.hist(df_employee['MonthlyIncome'], bins=30, color='lightblue', edgecolor='black')
plt.axvline(min_salary, color='red', linestyle='dashed', linewidth=1, label=f'Min: {min_salary}')
plt.axvline(max_salary, color='green', linestyle='dashed', linewidth=1, label=f'Max: {max_salary}')
plt.axvline(median_salary, color='orange', linestyle='dashed', linewidth=1, label=f'Median: {median_salary}')
plt.axvline(mean_salary, color='purple', linestyle='dashed', linewidth=1, label=f'Mean: {mean_salary:.2f}')
plt.axvline(mode_salary, color='cyan', linestyle='dashed', linewidth=1, label=f'Mode: {mode_salary}')
plt.title('Distribution of Monthly Income with Statistical Lines')
plt.xlabel('Monthly Income')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```
