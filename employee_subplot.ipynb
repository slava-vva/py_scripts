```python
import pandas as pd
df_employee = pd.read_csv('WA_Fn-UseC_-HR-Employee-Attrition.csv')
```

```python
import matplotlib.pyplot as plt

```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 6))
daily_rate_by_department = df_employee.groupby('Department')['DailyRate'].mean()
ax[0].bar(daily_rate_by_department.index, daily_rate_by_department.values, color='orange')
ax[0].set_xlabel('Department')
ax[0].set_ylabel('Daily Rate')
ax[0].set_title('Daily Rate by Department')
ax[0].tick_params(axis='x', rotation=45)
attrition_counts = df_employee['Attrition'].value_counts()
ax[1].pie(attrition_counts, labels=attrition_counts.index, autopct='%1.1f%%', startangle=90)
ax[1].set_title('Attrition Distribution')
fig.show()
```
