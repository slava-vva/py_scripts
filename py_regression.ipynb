```python
import pandas as pd

df = pd.read_csv("WA_Fn-UseC_-HR-Employee-Attrition.csv")  # or your dataset name
df.head()
df.to_excel("output.xlsx", index=False)
```

```python
cols = [
    'OverTime',
    'EnvironmentSatisfaction',
    'YearsAtCompany',
    'MonthlyIncome',
    'HourlyRate',
    'JobInvolvement',
    'JobSatisfaction'
]

```

```python
# Convert OverTime to numeric (Yes=1, No=0)
df['OverTime_num'] = df['OverTime'].map({'Yes': 1, 'No': 0})

corr = df[
    ['OverTime_num', 'EnvironmentSatisfaction', 'YearsAtCompany',
     'MonthlyIncome', 'HourlyRate', 'JobInvolvement', 'JobSatisfaction']
].corr(method='spearman')

# Only correlations with JobSatisfaction
print(corr['JobSatisfaction'])
```
