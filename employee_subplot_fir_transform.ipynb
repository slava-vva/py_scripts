```python
import pandas as pd
df_employee = pd.read_csv('WA_Fn-UseC_-HR-Employee-Attrition.csv')
```

```python
import matplotlib.pyplot as plt

```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df_normalized = df_employee.groupby('EducationField')[['DailyRate', 'MonthlyIncome']].mean().copy()

print(df_normalized)

# Normalize values between 0 and 1
df_normalized[['DailyRate', 'MonthlyIncome']] = scaler.fit_transform(
    df_normalized[['DailyRate', 'MonthlyIncome']]
)

print(df_normalized)

# Plot
df_normalized.plot(
    y=['DailyRate', 'MonthlyIncome'],
    kind='bar',
    figsize=(8, 5)
)
```
